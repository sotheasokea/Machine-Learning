```python
import matplotlib.pyplot as plt
```

```python
from sklearn.tree import DecisionTreeClassifier, plot_tree
```

```python
X = [[0, 0], [1, 1]]
Y = [0, 1]
X1 = [[0, 0], [1, -1]]
Y1 = [0, 1]

plt.plot(X, Y, marker = '*', label = 'xy')
plt.plot(X1, Y1, marker = 'o', label = 'x1y1')
plt.legend()
```

```python
X = [[0, 0], [1, 1], [2, 2]]
Y = [0, 1, 2]
clf = DecisionTreeClassifier(random_state = 1)
clf.fit(X, Y)

```

```python
plt.figure(figsize=(6, 4))

plot_tree(
    clf,
    # feature_names=["Feature 1", "Feature 2", "feature 3"],
    # class_names=["Class 0", "Class 1", "Class 2"],
    # filled=True,
    # rounded=True
)
plt.tight_layout()
plt.show()
```

```python
clf.predict([[-5, -4]])
```

```python
clf.predict_proba([[2, -4]])
```
