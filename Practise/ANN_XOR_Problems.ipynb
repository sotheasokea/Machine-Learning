```python
import numpy as np
from sklearn.neural_network import MLPClassifier
import matplotlib.pyplot as plt
```

```python
input = np.array([
  [0, 0],
  [0, 1],
  [1, 0],
  [1, 1]
])
target = np.array([0, 1, 1, 0])
print(input)
```

+ changing random_state will effect the model loss
+ having low max_iter, will cause our model to underfitting (cut off before it finished converging)

```python
model = MLPClassifier(hidden_layer_sizes=(4, ), random_state = 55, max_iter = 10000)
model.fit(input, target)
plt.plot(model.loss_curve_)
plt.xlabel("iteration")
plt.ylabel("loss")
plt.show()
```

```python
prediction = model.predict(input)
print(prediction)
```

```python
print(model.loss_)
```

# number of neurons in each hidden_layer 

+ Adding more layers/neurons increases model capacity
+ Have less layers : incapable of representing the pattern
+ More layers on small dataset: can make training harder

```python
model = MLPClassifier(hidden_layer_sizes=(2,2,2, ), random_state = 55, max_iter = 10000)
model.fit(input, target)
plt.plot(model.loss_curve_)
plt.xlabel("iteration")
plt.ylabel("loss")
plt.show()
```

```python
prediction = model.predict(input)
print(prediction)
print(model.loss_)
```
