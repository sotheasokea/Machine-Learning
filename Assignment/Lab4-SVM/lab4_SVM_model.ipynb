```python
import numpy as np
from sklearn.datasets import load_breast_cancer, load_digits
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, multilabel_confusion_matrix
import pandas as pd
from sklearn.preprocessing import StandardScaler
import dataframe_image as dfi
```

```python
# breath_cancer dataset
data = load_breast_cancer()
# print(data.data)
# print(data.target)
X = data.data
Y = data.target
```

```python
print(X.shape)
# print(Y) # Y has value 0, 1 so we use SVC - Support Vector Classification
```

```python
# 20% of data for test and 80% for training model
train_features, test_features, train_labels, test_labels = train_test_split(X, Y, test_size = 0.2, random_state = 55)
print(train_labels.shape)
print(train_features.shape)
```

```python
model_svc = SVC(kernel = 'rbf')
model_svc.fit(train_features, train_labels)

label_predicted = model_svc.predict(test_features)
print(classification_report(test_labels, label_predicted))
```

```python
# load_digits
data = load_digits()
features = data.data
labels = data.target
print(features.shape)
print(labels.shape)
```

```python
print(np.unique(labels))
print(features.min(), features.max())
```

```python
train_features, test_features, train_labels, test_labels = train_test_split(features, labels, test_size = 0.2, random_state = 55)
```

```python
scaler = StandardScaler()
train_features_scaled = scaler.fit_transform(train_features)
test_features_scaled = scaler.fit_transform(test_features)
```

```python
model = SVC(kernel = 'rbf')
model.fit(train_features_scaled, train_labels)

predicted = model.predict(test_features_scaled)
print(classification_report(test_labels, predicted))
```

```python
# cm = confusion_matrix(test_labels, predicted)
# print(cm.shape)
# tn, fp, fn, tp = cm.ravel()
#
# sensitivity = tp / (tp + fn) # all positive, how many we got
# specificity = tn / (tn + fp) # all negative, how many we got

#
# print(f"Sensitivity (Recall): {sensitivity:.4f}")
# print(f"Specificity:          {specificity:.4f}")
sensitivities = []
specificities = []
mcm = multilabel_confusion_matrix(test_labels, predicted)
for cm_i in mcm:
    tn, fp, fn, tp = cm_i.ravel()
    sensitivities.append(tp / (tp + fn))
    specificities.append(tn / (tn + fp))

avg_sensitivity = np.mean(sensitivities)
avg_specificity = np.mean(specificities)

print(f"Average Sensitivity: {avg_sensitivity:.4f}")
print(f"Average Specificity: {avg_specificity:.4f}")
```

```python
comparison_table = pd.DataFrame({
    'result': range(1, len(test_labels) + 1),   # 1 to 114
    'Ground Truth (Target)': test_labels,
    'SVM ที่ใช้ rbf เป็น kernel': predicted
})
print(confusion_matrix(test_labels, predicted))
display(comparison_table)
comparison_table.to_csv('digits_gt_vs_svm_comparison.csv', index=False)
```
