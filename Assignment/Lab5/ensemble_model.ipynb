```python
from sklearn.datasets import load_wine
```

```python
data = load_wine()
X = data.data
Y = data.target
print(X.shape)
print(Y.shape)
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size = 0.2, random_state  = 55)
print(x_train.shape)
print(y_train.shape)
```

```python
from sklearn.metrics import classification_report, accuracy_score, recall_score, f1_score
def model_evaluation(y_test, predict):
    print("acc:",accuracy_score(y_test, predict))
    print("f1:",f1_score(y_test, predict, average='macro'))
```

```python
from sklearn.tree import DecisionTreeClassifier
max_depth = [1, 2, 3, 4, 5, 10, 100, 1000]
for d in max_depth:
    print("max_depth:",d)
    dt_model = DecisionTreeClassifier(max_depth = d)
    dt_model.fit(x_train, y_train)
    predict = dt_model.predict(x_test)
    model_evaluation(y_test, predict)
```

```python
from sklearn.ensemble import RandomForestClassifier
max_depth = [1, 2, 3, 4, 5, 10, 100, 1000]
for d in max_depth:
    print("max_depth:",d)
    dt_model = RandomForestClassifier(max_depth = d)
    dt_model.fit(x_train, y_train)
    predict = dt_model.predict(x_test)
    model_evaluation(y_test, predict)
```

```python
from sklearn.ensemble import RandomForestClassifier
n_estimators = [1, 5, 10, 20, 50, 100, 200, 500]
for n in n_estimators:
    print("n_estimators:",n)
    dt_model = RandomForestClassifier(n_estimators = n, max_depth = 3)
    dt_model.fit(x_train, y_train)
    predict = dt_model.predict(x_test)
    model_evaluation(y_test, predict)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error
def model_evl(y_test, predict):
    print("mse:", mean_absolute_error(y_test, predict))
    print("r2:", r2_score(y_test, predict))
    # print("acc:", accuracy_score(y_test, predict))
```

```python
from sklearn.datasets import load_linnerud
from sklearn.multioutput import MultiOutputRegressor
data = load_linnerud()
X = data.data
Y = data.target
print(data.feature_names)
print(data.target_names)
# print(Y)
x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size = 0.2, random_state = 55)
print(x_train.shape)
# if we want to choose only one output
print(y_train[:, 0])
print(y_train)
# print(y_train)
```

```python
from sklearn.ensemble import GradientBoostingRegressor
learning_rate = [0.001, 0.005, 0.01, 0.05, 0.1, 0.2, 0.5, 1.0]
for lr in learning_rate:
    print("learning_rate:",lr)
    gb_model = MultiOutputRegressor(GradientBoostingRegressor(learning_rate = lr))
    gb_model.fit(x_train, y_train)
    predict = gb_model.predict(x_test)
    model_evl(y_test, predict)
```

```python
n_estimators = [1, 5, 10, 20, 50, 100, 200, 500]
for n in n_estimators:
    print("learning_rate:",n)
    gb_model = MultiOutputRegressor(GradientBoostingRegressor(n_estimators = n, learning_rate = 1.0))
    gb_model.fit(x_train, y_train)
    predict = gb_model.predict(x_test)
    model_evl(y_test, predict)
```
